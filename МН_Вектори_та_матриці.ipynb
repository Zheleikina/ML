## Завдання 3. Скалярний добуток і норма вектора

Для векторів `u = [3, -1, 2]` та `v = [1, 4, -2]`:

1. Обчисліть скалярний добуток.
2. Знайдіть евклідову норму кожного вектора.
3. Обчисліть косинус кута між векторами за формулою

\[
\cos(\alpha)=\frac{u\cdot v}{\|u\|\|v\|}
\]


````python
u = np.array([3, -1, 2])
v = np.array([1, 4, -2])

# Ваш код тут

dot_product = u @ v

norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)

cos_alpha = dot_product / (norm_u * norm_v)

print("Скалярний добуток:", dot_product)
print("Норма u:", norm_u)
print("Норма v:", norm_v)
print("Косинус кута:", cos_alpha)
````

## Завдання 8. Визначник та обернена матриця

Для матриці

\[
A =
\begin{bmatrix}
4 & 7\\
2 & 6
\end{bmatrix}
\]

1. Знайдіть визначник.
2. Знайдіть обернену матрицю.
3. Перевірте результат множенням `A @ A_inv`.
4. Поясніть у коментарі, який результат має бути отриманий.

````python
A = np.array([[4, 7],
              [2, 6]])

# Ваш код тут

det_A = np.linalg.det(A)
A_inv = np.linalg.inv(A)

print("Визначник матриці A:", det_A)

print("\nОбернена матриця A:")
print(A_inv)

print("\nПеревірка A @ A_inv:")
print(A @ A_inv)

# Результат множення має бути одиничною матрицею
````

## Завдання 9. Розв'язання системи лінійних рівнянь

## Завдання 12. Власна функція

Напишіть функцію `vector_stats(x)`, яка приймає числовий вектор і повертає:

- мінімум;
- максимум;
- середнє;
- медіану;
- стандартне відхилення.

Перевірте функцію на векторі:

`[4, 7, 1, 9, 3, 8, 5]`.

````python
def vector_stats(x):
    # Ваш код тут
    minimum = np.min(x)
    maximum = np.max(x)
    mean = np.mean(x)
    median = np.median(x)
    std = np.std(x)

    return minimum, maximum, mean, median, std


# Перевірка функції
test_vector = np.array([4, 7, 1, 9, 3, 8, 5])

result = vector_stats(test_vector)

print("Мінімум:", result[0])
print("Максимум:", result[1])
print("Середнє:", result[2])
print("Медіана:", result[3])
print("Стандартне відхилення:", result[4])
````

## Завдання 13. Функція з умовою

Створіть функцію

\[
f(x) =
\begin{cases}
x^2, & x < 0 \\
2x + 1, & x \geq 0
\end{cases}
\]

Застосуйте її до вектора значень від `-5` до `5`.

Спробуйте реалізувати функцію двома способами:
1. через звичайну функцію Python;
2. через `np.where()`.

````python
x = np.arange(-5, 6)

# Варіант 1 — звичайна функція Python
def f(x):
    if x < 0:
        return x ** 2
    else:
        return 2 * x + 1

result_1 = np.array([f(value) for value in x])

print("Варіант 1:")
print(result_1)


# Варіант 2 — np.where()
result_2 = np.where(x < 0, x ** 2, 2 * x + 1)

print("\nВаріант 2:")
print(result_2)
````

## Завдання 15. Підсумкове

Згенеруйте матрицю `5 × 5` із випадкових цілих чисел від `1` до `100`.

Виконайте такі дії:
1. Знайдіть середнє значення кожного рядка.
2. Знайдіть середнє значення кожного стовпця.
3. Замініть усі елементи, менші за загальне середнє матриці, на `0`.
4. Знайдіть рядок із найбільшою сумою.
5. Нормалізуйте початкову матрицю за формулою:

\[
x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}
\]

6. Напишіть функцію, яка виконує таку нормалізацію для довільної числової матриці.

````python
rng = np.random.default_rng(42)
matrix = rng.integers(1, 101, size=(5, 5))

print("Початкова матриця:")
print(matrix)

# Ваш код тут

# 1. Середнє значення кожного рядка
row_mean = matrix.mean(axis=1)
print("\nСереднє значення кожного рядка:")
print(row_mean)

# 2. Середнє значення кожного стовпця
column_mean = matrix.mean(axis=0)
print("\nСереднє значення кожного стовпця:")
print(column_mean)

# 3. Заміна елементів, менших за загальне середнє, на 0
overall_mean = matrix.mean()
matrix_modified = matrix.copy()
matrix_modified[matrix_modified < overall_mean] = 0

print("\nЗагальне середнє:", overall_mean)
print("\nМатриця після заміни:")
print(matrix_modified)

# 4. Рядок із найбільшою сумою
row_sums = matrix_modified.sum(axis=1)
max_row = np.argmax(row_sums)

print("\nСуми рядків:")
print(row_sums)
print("Індекс рядка з найбільшою сумою:", max_row)
print("Рядок із найбільшою сумою:")
print(matrix_modified[max_row])

# 5. Нормалізація початкової матриці
x_min = matrix.min()
x_max = matrix.max()

matrix_norm = (matrix - x_min) / (x_max - x_min)

print("\nНормалізована матриця:")
print(matrix_norm)

# 6. Функція нормалізації довільної матриці
def normalize_matrix(x):
    x_min = x.min()
    x_max = x.max()
    return (x - x_min) / (x_max - x_min)

normalized = normalize_matrix(matrix)

print("\nРезультат роботи функції:")
print(normalized)
````
