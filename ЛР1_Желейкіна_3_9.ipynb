### Завдання 27
Дано матриці:

```text
A = [[1, 2],
     [3, 4]]

B = [[2, 0],
     [1, 2]]
```

Обчисліть:
1. поелементний добуток `A * B`;
2. матричний добуток `A @ B`.

Поясніть у коментарі до коду, чим відрізняються ці операції.

````python
# Код студента
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [2, 0],
    [1, 2]
])

elementwise = A * B
matrix_product = A @ B

print("Поелементний добуток:")
print(elementwise)

print("Матричний добуток:")
print(matrix_product)

# A * B множить відповідні елементи матриць. A @ B виконує матричне множення: рядки A множаться на стовпці B.
````

### Завдання 28
Для матриці:

```text
A = [[4, 7],
     [2, 6]]
```

обчисліть:
- визначник;
- обернену матрицю.

Перевірте результат, помноживши початкову матрицю на обернену.

````python
# Код студента

A = np.array([
    [4, 7],
    [2, 6]
])

det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
check = A @ inv_A

print("Визначник:", det_A)
print("Обернена матриця:")
print(inv_A)
print("Перевірка:")
print(check)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента

A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

normalized = (A - mean) / std

print(normalized)
````

Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
# Код студента

sales = np.array([
    [120, 150, 170, 160],
    [80, 90, 100, 110],
    [200, 180, 210, 230]
])

product_totals = np.sum(sales, axis=1)
month_totals = np.sum(sales, axis=0)
best_product = np.argmax(product_totals) + 1
best_month = np.argmax(month_totals) + 1
product_means = np.mean(sales, axis=1)

print("Продажі кожного товару:", product_totals)
print("Продажі в кожному місяці:", month_totals)
print("Товар з найбільшими продажами:", best_product)
print("Місяць з найбільшими продажами:", best_month)
print("Середні продажі кожного товару:", product_means)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента

matrix = np.random.randint(1, 21, size=(5, 5))

main_diag_sum = np.trace(matrix)
secondary_diag_sum = np.trace(np.fliplr(matrix))
row_max = np.max(matrix, axis=1)
col_min = np.min(matrix, axis=0)

mean_value = np.mean(matrix)
modified_matrix = matrix.copy()
modified_matrix[modified_matrix < mean_value] = 0

print("Матриця:")
print(matrix)
print("Сума головної діагоналі:", main_diag_sum)
print("Сума побічної діагоналі:", secondary_diag_sum)
print("Максимум кожного рядка:", row_max)
print("Мінімум кожного стовпця:", col_min)
print("Матриця після заміни:")
print(modified_matrix)
````

### Завдання 38
Дано матрицю ознак трьох об’єктів:

```text
X = [[2, 4, 6],
     [1, 3, 5],
     [5, 7, 9]]
```

Обчисліть **евклідову відстань** між:
- першим і другим рядком;
- першим і третім рядком;
- другим і третім рядком.

Використайте операції NumPy та `np.linalg.norm()`.

````python
# Код студента

X = np.array([
    [2, 4, 6],
    [1, 3, 5],
    [5, 7, 9]
])

d_12 = np.linalg.norm(X[0] - X[1])
d_13 = np.linalg.norm(X[0] - X[2])
d_23 = np.linalg.norm(X[1] - X[2])

print("Відстань між 1 і 2 рядками:", d_12)
print("Відстань між 1 і 3 рядками:", d_13)
print("Відстань між 2 і 3 рядками:", d_23)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
# Код студента

A = np.array([
    [1, 2, 3, 4],
    [2, 4, 6, 8],
    [3, 1, 4, 2],
    [5, 2, 1, 3]
])

norms = np.linalg.norm(A, axis=1, keepdims=True)
normalized_A = A / norms

print("Норми рядків:")
print(norms)

print("Нормалізована матриця:")
print(normalized_A)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента

A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)
check = A @ x

print("Розв'язок:", x)
print("A @ x:", check)
print("b:", b)
print("Результати збігаються:", np.allclose(check, b))
````
